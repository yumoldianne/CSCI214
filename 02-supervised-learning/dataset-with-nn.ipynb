## Dataset with Neural Networks

```python
import pandas as pd
from torch.utils.data import Dataset, DataLoader
import torch
import torch.nn as nn
```

```python
df = pd.read_csv("titanic_train_minmax_scaled.csv")
```

```python
class TitanicDataset(Dataset):
  def __init__(self, df):
    self.df = df
    self.X = torch.tensor(df.drop(columns = ['Survived']).values, dtype = torch.float32)
    # self.y = torch.tensor(df['Survived'].values, dtype = torch.float32)
    labels = torch.tensor(df["Survived"].values, dtype=torch.long)
    self.y = torch.zeros((len(labels), 2), dtype=torch.float32)
    self.y[torch.arange(len(labels)), labels] = 1.0
  
  def __len__(self):
    self.len = len(self.df)
    return self.len
  
  def __getitem__(self, index):
    return (self.X[index], self.y[index])

dataset = TitanicDataset(df)
X, y = dataset[3]

print(X)
print(y)
```

```python
loader = DataLoader(dataset, batch_size = 5, shuffle = True)
for batch_x, batch_y in loader:
  print(batch_x.shape)
  print(batch_y.shape)
  break
```

```python
class TitanicNeuralNetwork(nn.Module):
    def __init__(self, input_size, output_size):
        super().__init__()
        self.input_size = input_size
        self.output_size = output_size

        self.relu = nn.ReLU()

        self.input_to_hidden = nn.Linear(self.input_size, 5)

        self.hidden_to_output = nn.Linear(5, self.output_size)

    def forward(self, x):
        x = self.input_to_hidden.forward(x)
        x = self.relu.forward(x)
        x = self.hidden_to_output.forward(x)
        x = self.relu.forward(x)

        return x

titanic_nn = TitanicNeuralNetwork(10, 2)
print(titanic_nn)
```
